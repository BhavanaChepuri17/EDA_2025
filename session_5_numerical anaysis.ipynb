```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

visa_df=pd.read_csv(r"C:\Users\jsamu\Documents\Naresh IT\Python\Visadataset.csv")
cat=visa_df.select_dtypes(include="object").columns
num=visa_df.select_dtypes(exclude="object").columns
```

```python
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.hist(wage_data,bins=40,color="skyblue",edgecolor="black")
plt.subplot(1,2,2)
plt.boxplot(wage_data,vert=False)
plt.show()
```

```python
q1=np.percentile(wage_data,25)
q2=np.percentile(wage_data,50)
q3=np.percentile(wage_data,75)
IQR=q3-q1
lb=q1-1.5*IQR
ub=q3+1.5*IQR
con1=wage_data>=lb
con2=wage_data<=ub
con=wage_data[con1&con2]
Median=np.median(con)
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.hist(Median,bins=30,color="skyblue",edgecolor="black")
plt.title("histogram of Non-outlier wages")
plt.xlabel("Wage")
plt.ylabel("Frequency")
plt.subplot(1,2,2)
plt.boxplot(con,vert=False)
plt.title("Boxplot of Non-outlier wage")
plt.show()
```

```python
q1=np.percentile(wage_data,25)
q2=np.percentile(wage_data,50)
q3=np.percentile(wage_data,75)
IQR=q3-q1
lb=q1-1.5*IQR
ub=q3+1.5*IQR
con1=wage_data>=lb
con2=wage_data<=ub
con=wage_data[con1&con2]
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.hist(con,bins=30,color="skyblue",edgecolor="black")
plt.axvline(q2,color="red",label="Median")
plt.title("histogram of Non-outlier wages")
plt.xlabel("Wage")
plt.ylabel("Frequency")
plt.subplot(1,2,2)
plt.boxplot(con,vert=False)
plt.title("Boxplot of Non-outlier wage")
plt.show()
```

```python
q1=np.percentile(wage_data,25)
q2=np.percentile(wage_data,50)
q3=np.percentile(wage_data,75)
IQR=q3-q1
lb=q1-1.5*IQR
ub=q3+1.5*IQR
con1=wage_data<lb
con2=wage_data>ub
wage_cleaned=np.where(con1|con2,q2,wage_data)
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.hist(wage_cleaned,bins=30,color="skyblue",edgecolor="black")
plt.title("histogram of outlier wages")
plt.xlabel("Wage")
plt.ylabel("Frequency")
plt.subplot(1,2,2)
plt.boxplot(wage_cleaned,vert=False)
plt.title("Boxplot of outlier wage")
plt.show()
```
